```python
import pandas as pd
import numpy as np


data = {
    'Student ID': ['S001', 'S002', 'S003', 'S004', 'S005',
                    'S006', 'S007', 'S008', 'S009', 'S010'],
    'Name': ['Aarav Sharma', 'Anisha Raut', 'Bibek Thapa', 'Sneha Gurung',
             'Rohan Karki', 'Priya Shrestha', 'Suman Adhikari', 'Aayusha Lama',
             'Nischal Rai', 'Kritika Bhandari'],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male',
               'Female', 'Male', 'Female', 'Male', 'Female'],
    'Science': [85, 92, 74, 89, 68, 95, 81, 77, 88, 91],
    'English': [78, 88, 81, 91, 72, 89, 75, 84, 79, 86],
    'Nepali': [82, 90, 76, 87, 70, 93, 79, 88, 85, 89],
    'Hobbies': ['Cricket', 'Reading', 'Football', 'Dancing', 'Gaming',
                'Painting', 'Photography', 'Singing', 'Cycling', 'Reading']
}

df = pd.DataFrame(data)
print("Step 1: Student DataFrame")
print(df)
marks = df[['Science', 'English', 'Nepali']].to_numpy()

print("\nStep 2: NumPy Properties (marks matrix)")
print("Shape:", marks.shape)
print("Size :", marks.size)
print("Ndim :", marks.ndim)
```

```python
print("\nStep 3: Operations")

print("\nTranspose (subjects as rows):\n", marks.transpose())

print("\nTotal marks per student (Science+English+Nepali):\n",
      marks.sum(axis=1))

print("\nAverage marks per subject:\n",
      marks.mean(axis=0))
```

```python
block_a = marks[:3]
block_b = marks[3:6]

print("\nAddition (block_a + block_b):\n", block_a + block_b)
print("\nSubtraction (block_a - block_b):\n", block_a - block_b)
print("\nElement-wise Multiplication:\n", block_a * block_b)
print("\nMatrix Product (block_a x block_b):\n", np.dot(block_a, block_b))

for name, block in [("block_a", block_a), ("block_b", block_b)]:
    det = np.linalg.det(block)
    print(f"\nDeterminant of {name}:", det)
    if abs(det) > 1e-9:
        print(f"Inverse of {name}:\n", np.linalg.inv(block))
    else:
        print(f"{name} is singular, inverse does not exist")
```

```python
print("\nOptional: One-Hot Encoding (Gender & Hobbies)")

encoded = pd.get_dummies(df, columns=['Gender', 'Hobbies']).astype(
    {col: int for col in pd.get_dummies(df, columns=['Gender', 'Hobbies']).columns
     if pd.get_dummies(df, columns=['Gender', 'Hobbies'])[col].dtype == bool}
)
print(encoded)
```
