```python
import pandas as pd
import numpy as np
from scipy import stats

pd.set_option("display.width", 100)

# ---------------------------------------------------------------
# Load data
# ---------------------------------------------------------------
url = "https://raw.githubusercontent.com/vincentarelbundock/Rdatasets/master/csv/AER/TeachingRatings.csv"
df = pd.read_csv(url)
df = df.drop(columns=["rownames"])
print("Shape:", df.shape)

mean_eval = df["eval"].mean()
std_eval = df["eval"].std()
n = len(df)
print(f"\nEvaluation score: mean = {mean_eval:.4f}, sd = {std_eval:.4f}, n = {n}")

# =================================================================
# Q1. P(eval > 4.5)
# =================================================================
print("\n" + "=" * 70)
print("Q1. PROBABILITY OF AN EVALUATION SCORE GREATER THAN 4.5")
print("=" * 70)

# (a) Theoretical -- assume eval ~ Normal(mean_eval, std_eval)
p_theoretical_q1 = 1 - stats.norm.cdf(4.5, loc=mean_eval, scale=std_eval)
z_q1 = (4.5 - mean_eval) / std_eval
print(f"Assuming eval ~ Normal(mean={mean_eval:.4f}, sd={std_eval:.4f}):")
print(f"z = (4.5 - {mean_eval:.4f}) / {std_eval:.4f} = {z_q1:.4f}")
print(f"P(eval > 4.5) = 1 - Phi({z_q1:.4f}) = {p_theoretical_q1:.4f}  ({p_theoretical_q1*100:.2f}%)")

# (b) Empirical -- directly from the data, for comparison
count_q1 = (df["eval"] > 4.5).sum()
p_empirical_q1 = count_q1 / n
print(f"\n(For comparison) Empirical, straight from the data:")
print(f"{count_q1} out of {n} instructors scored above 4.5 -> P = {p_empirical_q1:.4f} ({p_empirical_q1*100:.2f}%)")

# =================================================================
# Q2. P(3.5 < eval < 4.2)
# =================================================================
print("\n" + "=" * 70)
print("Q2. PROBABILITY OF AN EVALUATION SCORE BETWEEN 3.5 AND 4.2")
print("=" * 70)

z_low = (3.5 - mean_eval) / std_eval
z_high = (4.2 - mean_eval) / std_eval
p_theoretical_q2 = stats.norm.cdf(4.2, loc=mean_eval, scale=std_eval) - stats.norm.cdf(3.5, loc=mean_eval, scale=std_eval)
print(f"z_low  = (3.5 - {mean_eval:.4f}) / {std_eval:.4f} = {z_low:.4f}")
print(f"z_high = (4.2 - {mean_eval:.4f}) / {std_eval:.4f} = {z_high:.4f}")
print(f"P(3.5 < eval < 4.2) = Phi({z_high:.4f}) - Phi({z_low:.4f}) = {p_theoretical_q2:.4f}  ({p_theoretical_q2*100:.2f}%)")

count_q2 = ((df["eval"] > 3.5) & (df["eval"] < 4.2)).sum()
p_empirical_q2 = count_q2 / n
print(f"\n(For comparison) Empirical, straight from the data:")
print(f"{count_q2} out of {n} instructors scored between 3.5 and 4.2 -> P = {p_empirical_q2:.4f} ({p_empirical_q2*100:.2f}%)")

# =================================================================
# Q3. Two-tailed z-test -- regional players vs. historic pro mean
# =================================================================
print("\n" + "=" * 70)
print("Q3. TWO-TAILED Z-TEST -- REGIONAL PLAYERS vs. HISTORIC PRO MEAN")
print("=" * 70)

mu0 = 12          # historic (population) mean, points per game
sigma = 5.5       # historic (population) standard deviation
n_players = 36    # sample size
xbar = 10.7       # sample mean, points per game
alpha = 0.05

print("H0: The mean points scored by regional players is NOT different from the historic mean (mu = 12).")
print("H1: The mean points scored by regional players IS different from the historic mean (mu != 12).")

se = sigma / np.sqrt(n_players)
z_stat = (xbar - mu0) / se
p_value = 2 * (1 - stats.norm.cdf(abs(z_stat)))
z_critical = stats.norm.ppf(1 - alpha / 2)

print(f"\nStandard error = sigma / sqrt(n) = {sigma} / sqrt({n_players}) = {se:.4f}")
print(f"z = (xbar - mu0) / SE = ({xbar} - {mu0}) / {se:.4f} = {z_stat:.4f}")
print(f"Two-tailed p-value = {p_value:.4f}")
print(f"Critical z-value at alpha={alpha} (two-tailed) = +/-{z_critical:.4f}")

print(f"\nDecision (critical value method): |z| = {abs(z_stat):.4f} {'>' if abs(z_stat) > z_critical else '<='} z_critical = {z_critical:.4f}")
print(f"Decision (p-value method): p = {p_value:.4f} {'<' if p_value < alpha else '>='} alpha = {alpha}")

if abs(z_stat) > z_critical:
    print("=> REJECT H0: the regional players' mean score is significantly different from the historic mean.")
else:
    print("=> FAIL TO REJECT H0: not enough evidence that the regional players' mean score differs from the historic mean.")

print("\nDone.")
```
