```python
import pandas as pd
url = "https://raw.githubusercontent.com/sebastiansauer/Daten_Unterricht/master/TeachingRatings.csv"
df = pd.read_csv(url, sep=';', decimal=',')
df.head()
```

```python
df.shape          # should show (463, 12) for the standard dataset
df.info()
df.columns
```

```python
#question no.2
print("Mean:", df['students'].mean())
print("Median:", df['students'].median())
print("Min:", df['students'].min())
print("Max:", df['students'].max())
```

```python
#question no.3
df.describe()
```

```python
#question no.4
import matplotlib.pyplot as plt

plt.hist(df['beauty'], bins=15, edgecolor='black')
plt.xlabel('Beauty Score')
plt.ylabel('Frequency')
plt.title('Distribution of Beauty Scores')
plt.show()
```

```python
#question no.5
df.groupby('gender')['beauty'].agg(['mean', 'std'])
```
